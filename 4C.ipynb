```python
raw_text=open('document3.txt','r',encoding='utf-8').read()
```

```python
sentences=raw_text.split('\n')
```

```python
cleaned_sentences=[simple_preprocess(sentence) for sentence in sentences if sentence]
print(len(cleaned_sentences))
```

```python
model=Word2Vec(
    sentences=cleaned_sentences,
    vector_size=100,
    window=5,
    min_count=1,
    sg=0
)
```

```python
cleaned_sentences
```

```python
test_word='time'
if test_word in model.wv:
    print(f"words most mathematically similar to {test_word}:")
    similar_words=model.wv.most_similar(test_word,topn=4)
    for word,score in similar_words:
        print(f"- {word} and similarity is {score}")
    print(f"\nThe actual CBOW vector for '{test_word}' (first 5 numbers only):")
    print(model.wv[test_word][:5])
else:
    print(f"The word {test_word} was not found in docs")
```
